# Section 1 · Python 실습 환경 준비하기

Python, VS Code, Python·Jupyter 확장 프로그램과 Notebook 커널의 역할을 확인하고 첫 코드를 실행합니다.

**읽는 순서:** 설명을 읽은 뒤 코드 셀을 위에서부터 실행합니다. 이 파일에는 실제 실행 결과가 저장되어 있습니다. 저장된 버전·경로는 자료 제작 환경의 값이므로, 자신의 컴퓨터에서 실행하면 달라지는 것이 정상입니다.

| 구성 요소 | 역할 |
| --- | --- |
| Python | 코드를 실제로 계산하는 프로그램 |
| VS Code | 코드와 파일을 편집하는 작업 공간 |
| Python 확장 | Python 인터프리터 선택 및 코드 편집 지원 |
| Jupyter 확장 | `.ipynb`를 열고 코드 셀을 실행 |
| Notebook 커널 | 현재 노트북이 사용하는 Python 실행 과정 |

압축을 푼 `python_economics_notebooks` 폴더를 VS Code에서 열고, Microsoft의 Python·Jupyter 확장을 준비합니다. 오른쪽 위의 **커널 선택**에서 패키지를 설치한 Python을 고릅니다. `Shift+Enter`로 셀을 실행하고, 다시 검증할 때에는 **커널 다시 시작 → 모두 실행**을 사용합니다.

아래 터미널 명령은 필요한 설치 단계에서 직접 입력합니다. 운영체제에 맞는 명령 하나를 선택하며, 설치 명령은 ‘모두 실행’으로 자동 수행되지 않습니다.

## Python 설치하기

### Windows에서 설치하기

Windows 터미널에서 Python과 pip가 연결되는지 확인합니다.

*강의록: `section01_python_setup.tex` 146행 · 01-L001*

```shell
python --version
python -m pip --version
```

```text

```

</details>

Windows에서 `python` 명령이 잡히지 않을 때의 대안입니다.

*강의록: `section01_python_setup.tex` 166행 · 01-L003*

```shell
py --version

```

### macOS에서 설치하기

macOS에서는 `python3`로 확인합니다.

*강의록: `section01_python_setup.tex` 207행 · 01-L004*

```shell

```

<details><summary>강의록의 참고 출력 · 01-L005</summary>

```text

```

</details>

*강의록: `section01_python_setup.tex` 360행 · 01-L006*
