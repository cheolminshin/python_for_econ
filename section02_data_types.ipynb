# Section 2 · Python의 기본 자료형

정수, 실수, 문자열, 튜플, 사전, 부울을 직접 확인하고 금융 계산의 입력값을 적절한 자료형으로 바꿉니다.

**학습 순서:** 출력 결과를 먼저 예상한 뒤, 위에서부터 셀을 실행해 저장된 결과와 비교합니다. 이 노트북은 앞 장을 실행하지 않은 새 커널에서도 실행할 수 있습니다.

**오류 시연:** `TypeError`, `ValueError`, `KeyError`가 발생하는 예제는 해당 오류만 잡아서 실제 메시지를 출력합니다. 따라서 잘못된 입력이 왜 실패하는지 확인하면서도 ‘모두 실행’을 끝까지 진행할 수 있습니다.

연속형 수치는 `100.0`처럼 입력하고, 타입 주석과 실제 형변환을 구분합니다. `x: float = 100`이라고 적어도 값은 정수입니다.
